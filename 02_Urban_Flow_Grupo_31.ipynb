# Ejercicio 02

> puntos: 2

En este punto realizaremos verificaciones sobre el dataset de imágenes:

- Listar todas las imágenes disponibles (no mostrar/no imprimir). Se debe mostrar el nombre y su tamaño en kb.

- Separar las imágenes en 2 grupos,"completes" y "plates" y contar cuantas imágenes hay en cada grupo. Ayuda calcular la resolución promedio. Al trabajar con la información de las imágenes vamos a utilizar un diccionario (denominado group_images) para almacenar información imortante que podemos utilizar más adelante.

La salida es
```
El grupo 'plates' contiene XX imágenes.
El grupo 'completes' contiene XX imágenes.
```

El diccionario es (guardarlo en `ubran_flow/data/interim/group_images.json`):
```
group_images = {
  'plates': [{
    'filename': contiene solo el nombre del archivo y la extensión,
    'width': contiene el ancho en pixeles,
    'height': contiene el alto en pixeles,
    'area': cálculo del área de la imagen,
    'path': contiene el path completo de la imagen original,
    'patent': va a contener la patente
  }],
  'completes': [{
    'filename': contiene solo el nombre del archivo y la extensión,
    'width': contiene el ancho en pixeles,
    'height': contiene el alto en pixeles,
    'area': cálculo del área de la imagen,
    'path': contiene el path completo de la imagen original,
    'patent': va a contener la patente
  }],
}
```

- Desarrollar un función para mostrar 8 imágenes de forma aleatorea en forma de tabla de 2 columnas y 4 filas. Debe mostrar 4 imágenes de cada grupo. Es una función reutilizable. En este punto se deben mostrar las imágenes originales. Cada imágen debe tener un ancho de 10 pulgas y un alto de 8 pulgadas, se debe mostrar el nombre de la imagen.



````python
#@title Ejercicio 02: listado y separación de imágenes en grupos
group_images: dict = {"plates": [], "completes": []}

archivos: list = [
  f for f in os.listdir(IMGS_RAW_DIR)
  if f.lower().endswith('.jpg')
]

for archivo in sorted(archivos):
  ruta: str = os.path.join(IMGS_RAW_DIR, archivo)
  tamanio_kb: float = round(os.path.getsize(ruta) / 1024, 2)
  img: np.ndarray = cv2.imread(ruta)
  alto: int = img.shape[0]
  ancho: int = img.shape[1]
  area: int = alto * ancho

  entrada: dict = {
    "filename": archivo,
    "width": ancho,
    "height": alto,
    "area": area,
    "size_kb": tamanio_kb,
    "path": ruta,
    "patent": ""
  }

  if "_patente" in archivo:
    group_images["plates"].append(entrada)
  else:
    group_images["completes"].append(entrada)

n_plates: int = len(group_images['plates'])
n_completes: int = len(group_images['completes'])
print(f"El grupo 'plates' contiene {n_plates} imágenes.")
print(f"El grupo 'completes' contiene {n_completes} imágenes.")
````

````python
#@title Ejercicio 02: guardar group_images en JSON
json_path: str = "urban_flow/data/interim/group_images.json"

with open(json_path, "w", encoding="utf-8") as f:
  json.dump(group_images, f, ensure_ascii=False, indent=2)

print(f"group_images guardado en {json_path}")
````

````python
#@title Ejercicio 02: función para mostrar imágenes en grilla
def mostrar_imagenes_grilla(
  plates_paths: list,
  completes_paths: list
) -> None:
  """Muestra 8 imágenes aleatorias en grilla de 2 columnas y 4 filas,
  4 de cada grupo. Cada imagen ocupa 10x8 pulgadas."""
  seleccion: list = (
    random.sample(plates_paths, 4) +
    random.sample(completes_paths, 4)
  )

  fig, axes = plt.subplots(4, 2, figsize=(20, 32))
  axes = axes.flatten()

  for i, ruta in enumerate(seleccion):
    img: np.ndarray = cv2.imread(ruta)
    img_rgb: np.ndarray = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    axes[i].imshow(img_rgb)
    axes[i].set_title(os.path.basename(ruta), fontsize=8)
    axes[i].axis("off")

  plt.tight_layout()
  plt.show()


plates_paths: list = [img["path"] for img in group_images["plates"]]
completes_paths: list = [img["path"] for img in group_images["completes"]]

mostrar_imagenes_grilla(plates_paths, completes_paths)
````

````python
#@title Ejercicio 03: creación de estructura de carpetas
carpetas: list = [
  "urban_flow/data/interim/imgs/03_01_gray_scale/plates",
  "urban_flow/data/interim/imgs/03_01_gray_scale/completes",
  "urban_flow/data/interim/imgs/03_02_blur/plates",
  "urban_flow/data/interim/imgs/03_02_blur/completes",
  "urban_flow/data/interim/imgs/03_03_canny/plates",
  "urban_flow/data/interim/imgs/03_03_canny/completes",
]

for carpeta in carpetas:
  if not os.path.exists(carpeta):
    os.makedirs(carpeta)
    print(f"Creada: {carpeta}")
  else:
    print(f"Ya existe: {carpeta}")
````

````python
#@title Ejercicio 03: conversión a escala de grises
def procesar_gray_scale(
  group_images: dict,
  base_path: str
) -> dict:
  """Convierte las imágenes originales a escala de grises y las
  guarda en el directorio correspondiente según su grupo.

  Args:
    group_images: diccionario con grupos plates y completes.
    base_path: ruta base donde se guardan las imágenes procesadas.

  Returns:
    dict con las rutas de las imágenes procesadas por grupo.
  """
  rutas_procesadas: dict = {"plates": [], "completes": []}

  for grupo in ["plates", "completes"]:
    for imagen in group_images[grupo]:
      img: np.ndarray = cv2.imread(imagen["path"])
      gray: np.ndarray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
      destino: str = os.path.join(
        base_path, grupo, imagen["filename"]
      )
      cv2.imwrite(destino, gray)
      rutas_procesadas[grupo].append(destino)

  return rutas_procesadas


gray_paths: dict = procesar_gray_scale(
  group_images,
  "urban_flow/data/interim/imgs/03_01_gray_scale"
)

print(f"Plates procesadas: {len(gray_paths['plates'])}")
print(f"Completes procesadas: {len(gray_paths['completes'])}")
````

````python
#@title Ejercicio 03: suavizado con GaussianBlur
def procesar_blur(
  gray_paths: dict,
  base_path: str
) -> dict:
  """Aplica suavizado GaussianBlur sobre las imágenes en escala de
  grises y las guarda según su grupo.

  Args:
    gray_paths: diccionario con rutas de imágenes en escala de grises.
    base_path: ruta base donde se guardan las imágenes procesadas.

  Returns:
    dict con las rutas de las imágenes suavizadas por grupo.

  Parámetros ajustables:
    kernel_size (5, 5): tamaño de la ventana de suavizado.
      Aumentar para mayor suavizado (ej: (9,9)),
      disminuir para preservar más detalle (ej: (3,3)).
    sigmaX 0: calculado automáticamente según el kernel.
  """
  rutas_procesadas: dict = {"plates": [], "completes": []}

  for grupo in ["plates", "completes"]:
    for ruta in gray_paths[grupo]:
      img: np.ndarray = cv2.imread(ruta, cv2.IMREAD_GRAYSCALE)
      blur: np.ndarray = cv2.GaussianBlur(img, (5, 5), 0)
      filename: str = os.path.basename(ruta)
      destino: str = os.path.join(base_path, grupo, filename)
      cv2.imwrite(destino, blur)
      rutas_procesadas[grupo].append(destino)

  return rutas_procesadas


blur_paths: dict = procesar_blur(
  gray_paths,
  "urban_flow/data/interim/imgs/03_02_blur"
)

print(f"Plates procesadas: {len(blur_paths['plates'])}")
print(f"Completes procesadas: {len(blur_paths['completes'])}")
````

````python
#@title Ejercicio 03: detección de bordes con Canny
def procesar_canny(
  blur_paths: dict,
  base_path: str
) -> dict:
  """Aplica detección de bordes Canny sobre las imágenes suavizadas
  y las guarda según su grupo.

  Args:
    blur_paths: diccionario con rutas de imágenes suavizadas.
    base_path: ruta base donde se guardan las imágenes procesadas.

  Returns:
    dict con las rutas de las imágenes con bordes detectados por grupo.

  Parámetros ajustables:
    threshold1 (100): umbral inferior para detección de bordes.
      Disminuir para detectar más bordes (ej: 50).
    threshold2 (200): umbral superior para detección de bordes.
      Disminuir para detectar más bordes (ej: 150).
  """
  rutas_procesadas: dict = {"plates": [], "completes": []}

  for grupo in ["plates", "completes"]:
    for ruta in blur_paths[grupo]:
      img: np.ndarray = cv2.imread(ruta, cv2.IMREAD_GRAYSCALE)
      canny: np.ndarray = cv2.Canny(img, 100, 200)
      filename: str = os.path.basename(ruta)
      destino: str = os.path.join(base_path, grupo, filename)
      cv2.imwrite(destino, canny)
      rutas_procesadas[grupo].append(destino)

  return rutas_procesadas


canny_paths: dict = procesar_canny(
  blur_paths,
  "urban_flow/data/interim/imgs/03_03_canny"
)

print(f"Plates procesadas: {len(canny_paths['plates'])}")
print(f"Completes procesadas: {len(canny_paths['completes'])}")
````

````python
# @title Definición de la función extraer_patente

import easyocr

def extraer_patente(image_path: str) -> str:
  """Extrae la patente de una imagen usando OCR.

  Args:
    image_path: Ruta de la imagen a procesar.

  Returns:
    Patente extraída como string en mayúsculas,
    o cadena vacía si no se detecta texto.
  """
  reader = easyocr.Reader(['en'], gpu=False)
  detections = reader.readtext(image_path, detail=1)

  if len(detections) == 0:
    return ""

  # Tomar el texto con mayor confianza
  detections_sorted = sorted(
    detections,
    key=lambda d: d[2],
    reverse=True
  )
  texto = detections_sorted[0][1]

  # Limpiar: solo alfanuméricos, uppercase
  patente = ''.join(
    c for c in texto if c.isalnum()
  ).upper()

  return patente
````

````python
# @title Extracción de patentes con OCR

JSON_PATH: str = "urban_flow/data/interim/group_images.json"

# Cargar group_images desde JSON (por si se reinició la sesión)
with open(JSON_PATH, "r", encoding="utf-8") as f:
  group_images = json.load(f)

# Aplicar OCR a cada imagen del grupo plates
for imagen in group_images["plates"]:
  imagen["patent"] = extraer_patente(imagen["path"])

# Guardar JSON actualizado con patentes extraídas
with open(JSON_PATH, "w", encoding="utf-8") as f:
  json.dump(group_images, f, ensure_ascii=False, indent=2)

print(f"Patentes extraídas: {len(group_images['plates'])}")
````

````python
# @title Función de similitud de patentes

def calcular_ratio(patente_ocr: str, patente_csv: str) -> float:
  """Calcula ratio de similitud entre dos patentes comparando
  carácter a carácter de izquierda a derecha.

  Args:
    patente_ocr: Patente extraída por OCR.
    patente_csv: Patente del registro de multas.

  Returns:
    Float entre 0.0 y 1.0 representando la similitud.
  """
  if not patente_csv:
    return 0.0

  a: str = patente_ocr.upper().strip()
  b: str = patente_csv.upper().strip()

  coincidencias: int = sum(
    1 for i in range(min(len(a), len(b))) if a[i] == b[i]
  )

  return round(coincidencias / len(b), 4)
````

````python
# @title Matching patentes OCR vs CSV

FINES_PATH: str = "urban_flow/data/interim/speeding_fines.csv"
UMBRAL: float = 0.80

def matching_patentes(
  group_images: dict,
  fines_path: str,
  umbral: float
) -> pd.DataFrame:
  """Cruza registros de multas con imágenes OCR.

  Toma todas las multas como base y agrega imagen,
  patente_imagen y ratio cuando existe match >= umbral.

  Args:
    group_images: Diccionario con imágenes y patentes OCR.
    fines_path: Ruta al CSV de multas procesado.
    umbral: Ratio mínimo para considerar match válido.

  Returns:
    DataFrame completo de multas con columnas imagen,
    patente_imagen y ratio agregadas.
  """
  df_multas: pd.DataFrame = pd.read_csv(fines_path)

  imagenes_col: list = []
  patente_imagen_col: list = []
  ratio_col: list = []

  for _, fila in df_multas.iterrows():
    patente_csv: str = str(fila["patente"]) if pd.notna(
      fila["patente"]
    ) else ""
    mejor_ratio: float = 0.0
    mejor_imagen: str = ""
    mejor_patente_ocr: str = ""

    for imagen in group_images["plates"]:
      patente_ocr: str = imagen.get("patent", "")
      ratio: float = calcular_ratio(patente_ocr, patente_csv)
      if ratio > mejor_ratio:
        mejor_ratio = ratio
        mejor_imagen = imagen["filename"]
        mejor_patente_ocr = patente_ocr

    if mejor_ratio >= umbral:
      imagenes_col.append(mejor_imagen)
      patente_imagen_col.append(mejor_patente_ocr)
      ratio_col.append(mejor_ratio)
    else:
      imagenes_col.append("")
      patente_imagen_col.append("")
      ratio_col.append("")

  df_multas["imagen"] = imagenes_col
  df_multas["patente_imagen"] = patente_imagen_col
  df_multas["ratio"] = ratio_col

  return df_multas


df_resultado: pd.DataFrame = matching_patentes(
  group_images, FINES_PATH, UMBRAL
)

print(f"Total multas: {len(df_resultado)}")
print(f"Con imagen: {(df_resultado['imagen'] != '').sum()}")
print(df_resultado.head(10))
````

## Visualización de código

````python
from IPython.display import display, HTML, Code
import os

display(HTML("""
<style>
  .jp-RenderedText pre, .jp-RenderedHTMLCommon pre, .p-Widget pre {
    background-color: #1e1e1e !important;
    color: #d4d4d4 !important;
    border: 1px solid #FFF !important;
  }
  .highlight  { background: #1e1e1e !important; }
  .highlight pre { border: 2px solid !important; }
</style>
"""))
!rm -rf /content/sample_data .config
def visualizar_archivos_dark(directorio):
    for raiz, _, archivos in os.walk(directorio):
      if raiz.lower().endswith(('.git','.dvc', '__pycache__')):
        continue
      for nombre in archivos:
        if nombre.lower().endswith(('.py')):
          ruta = os.path.join(raiz, nombre)
          print(f"📁 {ruta}")
          with open(ruta, 'r') as f:
              contenido = f.read()
          display(Code(contenido, language=nombre.split('.')[-1]))
        if nombre.lower().endswith(('.md', '.csv')):
          ruta = os.path.join(raiz, nombre)
          print(f"📁 {ruta}")
          with open(ruta, 'r') as f:
              contenido = f.read()
          print("".join(contenido))
          print("\n\n")
          print("-" * 60)
          print("\n\n")

visualizar_archivos_dark('/content/')
````

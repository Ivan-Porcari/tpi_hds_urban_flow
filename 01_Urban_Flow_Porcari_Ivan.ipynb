## Ejercicio 04

> Puntos: 2

Definir una clase FineAnalyzer que cumpla con las siguientes funcionalidades:

- Reciba el DataFrame limpio como argumento de inicialización.

- Encapsule correctamente los datos.

- Implemente métodos para:

  - Ranking de las patentes más multadas (top 5). Debe retornar un DataFrame ordenado de mayor a menor y el índice debe comenzar desde 1, con las columnas de patentes y cantidad..

  - Ranking de los horarios donde se producen más multas (top 5). Debe retornar un DataFrame ordenado de mayor a menor y el índice debe comenzar desde 1, con las columnas de hora y cantidad.


  - Exceso promedio de velocidad. Debe retorna un flotante.

  - Exceso real promedio de velocidad . Debe retorna un flotante.

  - Contabilizar la cantidad de multas por cada  ubicación. Debe retornar un DataFrame ordenado por ubicación mostrando la ubicación y la cantidad.

- Crear el objecto e invocar cada uno de los métodos en celdas separadas.
  

````python
#FineAnalyzer - Definicion de clase
class FineAnalyzer:
  """Analizador de multas por exceso de velocidad."""

  def __init__(self, df: pd.DataFrame) -> None:
    """Inicia el analizador con el DataFrame limpio."""
    self.__df = df

  def ranking_patentes(self) -> pd.DataFrame:
    """Devuelve top 5 patentes más multadas, índice desde 1."""
    resultado = (
      self.__df['patente']
      .value_counts()
      .head(5)
      .reset_index()
    )
    resultado.columns = ['patente', 'cantidad']
    resultado.index = range(1, len(resultado) + 1)
    return resultado

  def ranking_horarios(self) -> pd.DataFrame:
    """Devuelve top 5 horarios con más multas, índice desde 1."""
    resultado = (
      self.__df['hora']
      .value_counts()
      .head(5)
      .reset_index()
    )
    resultado.columns = ['hora', 'cantidad']
    resultado.index = range(1, len(resultado) + 1)
    return resultado

  def exceso_promedio(self) -> float:
    """Devuelve el promedio de exceso de velocidad."""
    return float(self.__df['exceso_velocidad'].mean())

  def exceso_real_promedio(self) -> float:
    """Devuelve el promedio de exceso de velocidad real."""
    return float(self.__df['exceso_velocidad_real'].mean())

  def multas_por_ubicacion(self) -> pd.DataFrame:
    """Devuelve cantidad de multas por ubicación, ordenado."""
    resultado = (
      self.__df.groupby('ubicacion')
      .size()
      .reset_index(name='cantidad')
      .sort_values('ubicacion')
    )
    return resultado
````

````python
#FineAnalyzer - Carga de datos y creacion del objeto
df_limpio: pd.DataFrame = pd.read_csv(
  'urban_flow/data/interim/speeding_fines.csv'
)
analizador: FineAnalyzer = FineAnalyzer(df_limpio)
````

````python
#FineAnalyzer - Ranking de patentes mas multadas
analizador.ranking_patentes()
````

````python
#FineAnalyzer - Ranking de horarios con mas multas
analizador.ranking_horarios()
````

````python
#FineAnalyzer - Exceso promedio de velocidad
analizador.exceso_promedio()
````

````python
#FineAnalyzer - Exceso real promedio de velocidad
analizador.exceso_real_promedio()
````

````python
#FineAnalyzer - Multas por ubicacion
analizador.multas_por_ubicacion()
````

````python
#Ranking top 10 patentes reincidentes
top_patentes: pd.DataFrame = (
  df_limpio['patente']
  .value_counts()
  .head(10)
  .reset_index()
)
top_patentes.columns = ['patente', 'cantidad']

fig, ax = plt.subplots(figsize=(10, 6))
ax.bar(top_patentes['patente'], top_patentes['cantidad'])
ax.set_title('Top 10 patentes más reincidentes')
ax.set_xlabel('Patente')
ax.set_ylabel('Cantidad de multas')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('urban_flow/data/interim/plots/fines.jpg')
plt.show()
````

````python
#% de infracciones por hora
por_hora: pd.Series = (
  df_limpio['hora']
  .str[:2]  # toma solo "HH" de "HH:MM"
  .value_counts()
  .sort_index()
)

fig, ax = plt.subplots(figsize=(10, 8))
ax.pie(
  por_hora,
  labels=[f"{h}:00" for h in por_hora.index],
  autopct='%1.1f%%',
  startangle=90
)
ax.set_title('Porcentaje de infracciones por hora')
plt.tight_layout()
plt.savefig('urban_flow/data/interim/plots/hours.jpg')
plt.show()
````

````python
#Infracciones por mes
df_limpio['mes'] = pd.to_datetime(
  df_limpio['fecha'], errors='coerce'
).dt.month
por_mes: pd.Series = (
  df_limpio['mes']
  .value_counts()
  .sort_values(ascending=True)
)
nombres_mes: list[str] = [
  'Ene','Feb','Mar','Abr','May','Jun',
  'Jul','Ago','Sep','Oct','Nov','Dic'
]
etiquetas: list[str] = [
  nombres_mes[m - 1] for m in por_mes.index
]

fig, ax = plt.subplots(figsize=(8, 6))
ax.barh(etiquetas, por_mes.values)
ax.set_title('Infracciones por mes')
ax.set_xlabel('Cantidad')
plt.tight_layout()
plt.savefig('urban_flow/data/interim/plots/months.jpg')
plt.show()
````

````python
print("Filas con hora 00:00:", 
  df_limpio[df_limpio['hora'] == '00:00'].shape[0])
print("Filas con fecha 1932-01-01:", 
  df_limpio[df_limpio['fecha'] == '1932-01-01'].shape[0])
````

````python
#Excesos de velocidad en hora 00:00
df_hora: pd.DataFrame = df_limpio[
  (df_limpio['hora'] == '00:00') &
  (df_limpio['fecha'] != '1932-01-01')
].copy()
df_hora['fecha_dt'] = pd.to_datetime(
  df_hora['fecha'], errors='coerce'
)
df_hora['exceso_velocidad'] = pd.to_numeric(
  df_hora['exceso_velocidad'], errors='coerce'
)
excesos_hora: pd.Series = (
  df_hora.groupby('fecha_dt')['exceso_velocidad'].mean()
  .sort_index()
)

fig, ax = plt.subplots(figsize=(12, 5))
ax.plot(excesos_hora.index, excesos_hora.values)
ax.set_title('Exceso de velocidad promedio - hora 00:00')
ax.set_xlabel('Fecha')
ax.set_ylabel('Exceso promedio')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('urban_flow/data/interim/plots/hour.jpg')
plt.show()
````

````python
#Excesos de velocidad en fecha 1932-01-01
df_fecha: pd.DataFrame = df_limpio[
  df_limpio['fecha'] == '1932-01-01'
].copy()
df_fecha['exceso_velocidad'] = pd.to_numeric(
  df_fecha['exceso_velocidad'], errors='coerce'
)
excesos_fecha: pd.Series = (
  df_fecha.groupby('hora')['exceso_velocidad'].mean()
  .sort_index()
)

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(excesos_fecha.index, excesos_fecha.values)
ax.set_title('Exceso de velocidad promedio - fecha 1932-01-01')
ax.set_xlabel('Hora')
ax.set_ylabel('Exceso promedio')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('urban_flow/data/interim/plots/date.jpg')
plt.show()
````

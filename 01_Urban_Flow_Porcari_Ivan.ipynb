##### Normalizar Fecha 

````python
# Convierte la columna a fecha, si los datos son invalidos, los 
# transforma en NaT para no generar un error
#
  
df['fecha'] = pd.to_datetime(df['fecha'], errors='coerce', format='mixed')

df['fecha'] = df['fecha'].fillna('1932-01-01') # Podria usar pd.Timestamp('1900-01-01)

print(df['fecha'].head(10))

````

##### Normalizar Hora

````python
# Convierte la columna a hora, busca horas en diferentes formatos y 
# los incorrectos los convierte en nulos
df['hora'] = pd.to_datetime(df['hora'], errors='coerce', format='mixed')

# Aplica a las horas un formato de 24hs sin segundos
# Rellena los nulos con 00:00 
df['hora'] = df['hora'].dt.strftime('%H:%M').fillna('00:00')
print(df['hora'].head(10))
````

##### Funcion de limpieza de ubicaciones


````python
def normalizar_ubicaciones(texto:str):

    if pd.isna(texto):
        return "Desconocida"
    
    # Buscar 'av' o 'av.' (ignora mayúsculas/minúsculas)
    # av -> busca letras a y v
    # \. -> quiero que el punto actue como punto
    texto = re.sub(r'(?i)av\.?\s*', 'AV ', texto)
    
    # Buscar cualquier cosa que no sea letra, número o espacio.
    texto = re.sub(r'[^a-zA-Z0-9\s]', '', texto)

    # Buscar espacios en blanco 
    texto= re.sub(r'\s+', ' ', texto).strip()

    #Retorno palabra en Mayusculas
    return texto.upper()
````

##### Mostrar las primera 10 ubicaciones

````python
# Aplicamos la función a la columna
df['ubicacion'] = df['ubicacion'].apply(normalizar_ubicaciones)

# Mostrar las primeras 10  ubicaciones
print(df['ubicacion'].head(10))
````

##### Limpieza y normalización de patentes


````python
def normalizar_patentes(valor):

    # Si el valor es nulo de entrada, devolvemos pd.NA 
    if pd.isna(valor) or valor == "":
        return pd.NA
    
     # Quitar caracteres especiales
    texto_limpio = re.sub(r'[^a-zA-Z0-9]', '', valor)

    # Verificar si después de la limpieza quedó vacío (caso de error)
    if texto_limpio == "":
        return pd.NA

    # Retornar en mayusculas
    return texto_limpio.upper()
````

````python
#Ejecutar la limpieza y normalizacion como string
df['patente'] = df['patente'].apply(normalizar_patentes).astype("string")
````

##### Eliminacion de filas con valores relevantes para las multas

````python
# Columnas con datos relevantes para generar multas
columnas_relevantes = [
    'patente', 
    'fecha', 
    'hora', 
    'velocidad_registrada', 
    'velocidad_maxima', 
    'ubicacion'
]
filas_totales_df= len(df)

# Se comentaron las siguientes dos lineas ya que si bien son datos relevantes 
# a la hora de realizar una multa, pero posteriormente se pide un gráfico soicitando
# estos datos
#df['fecha'] = df['fecha'].replace('1932-01-01', np.nan)
#df['hora'] = df['hora'].replace('00:00', np.nan)


# Solo buscamos en las columnas relevantes (subset), no en todo el df
# Eliminamos las filas con valores vacios
df = df.dropna(subset=columnas_relevantes)

# Restamos las filas actuales a las filas iniciales para saber 
# cuantas filas fueron eliminadas
filas_eliminadas = filas_totales_df - len(df)


print(f"Se eliminaron {filas_eliminadas} filas. Las columnas observadas son:")
for columna in columnas_relevantes:
    print(f"- {columna}")
````

##### Detectar y eliminar outliers

````python
# Definimos las columnas numéricas donde buscaremos los outliers
columnas_observadas = ['velocidad_registrada', 'velocidad_maxima']

# Guardo las filas antes de limpiar
filas_antes = len(df)

#  Aplicamos el método del rango intercuartilico (IQR) para cada columna
for columna in columnas_observadas:
    # Calculamos el primer cuartil (Q1) 
    # Calculamos el tercer cuartil (Q3) 
    # Calculamos el rango intercuartílico

    Q1 = df[columna].quantile(0.25)    
    Q3 = df[columna].quantile(0.75)    
    IQR = Q3 - Q1
    
    # Definimos los limites permitidos
    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR
    
    # Conservamos solo las filas que esten dentro de los limites
    df = df[(df[columna] >= limite_inferior) & (df[columna] <= limite_superior)]

# Calculamos matemáticamente cuántas filas se eliminaron
filas_eliminadas = filas_antes - len(df)


print(f"Se eliminaron {filas_eliminadas} filas. Las columnas observadas son:")
for col in columnas_observadas:
    print(f"- {col}")
````

##### Exceso de velocidad real

````python
# Creo la columna con la resta de la velocidad registrada y la velocidad maxima
df['exceso_velocidad_real'] = df['velocidad_registrada'] - df['velocidad_maxima']

# Seleccionamos solo las columnas patente y exceso_velocidad_real y tomamos las primeras 10
print(df[['patente', 'exceso_velocidad_real']].head(10))
````

##### Exceso de velocidad

````python
# Agregamos el 5% de velocidad a la velocidad maxima
velocidad_tolerada = df['velocidad_maxima'] * 1.05

# Creamos la columna exceso de velocidad y hacemos la resta
df['exceso_velocidad'] = df['velocidad_registrada'] - velocidad_tolerada

# 3. Filtramos las columnas pedidas y mostramos las 10 primeras
print(df[['patente', 'exceso_velocidad']].head(10))
````

##### Eliminando las filas sin infracciones

````python
# Guardamos la cantidad total de filas que hay en la tabla actualmente
filas_antes = len(df)

# Conservamos las filas con exceso de velocidad mayor a 0
df = df[df['exceso_velocidad'] > 0]

# Calculamos las filas eliminadas
filas_eliminadas = filas_antes - len(df)

print(f"Se eliminaron {filas_eliminadas} filas.")
````

##### Porcentaje de infracciones con fecha 1932-01-01

Se calcula el porcentaje de infracciones registradas con la fecha 
1932-01-01. Esta fecha fue asignada en el Ejercicio 03 a todos los 
registros que tenían una fecha inválida en el sistema heredado. Se 
divide la cantidad de filas con ese valor sobre el total de filas 
del dataset limpio.

````python
# Calculo la cantidad total de registros del dataset df
# Esto representa el 100% de las multas registradas
total_filas = len(df)

# Calculo la cantidad de registros con fecha 1932-01-01
filas_fecha_invalida = len(df[df['fecha'] == '1932-01-01'])

# Regla de 3 simple:
# total_filas          -> 100%
# filas_fecha_invalida -> x
porcentaje_fecha = (filas_fecha_invalida * 100) / total_filas

print(f"El porcentaje de infracciones en la fecha 1932-01-01 es {porcentaje_fecha:.2f}%")
````

##### Porcentaje de infracciones con hora 00:00

Se calcula el porcentaje de infracciones registradas con la hora 
00:00. Este valor fue asignado en el Ejercicio 03 a todos los 
registros que tenían una hora inválida en el sistema heredado. Se 
divide la cantidad de filas con ese valor sobre el total de filas 
del dataset limpio.

````python
# Calculo la cantidad total de registros del dataset
# Esto representa el 100% de las multas registradas
total_filas = len(df)

# Calculo la cantidad de registros con hora 00:00
filas_hora_invalida = len(df[df['hora'] == '00:00'])

# Regla de 3 simple:
# total_filas        -> 100%
# filas_hora_invalida -> x
porcentaje_hora = (filas_hora_invalida * 100) / total_filas

print(f"El porcentaje de infracciones a la hora 00:00 es {porcentaje_hora:.2f}%")
````

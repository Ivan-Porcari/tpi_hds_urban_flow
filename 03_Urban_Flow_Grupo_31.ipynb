### Entidades y atributos

**Vehiculo**

**Radar**

**Multa**

| Atributo              | Tipo    | Descripción                        |
|-----------------------|---------|------------------------------------|
| multa_id              | INTEGER | Identificador único (PK)           |
| fecha                 | DATE    | Fecha de la infracción             |

| velocidad_registrada  | FLOAT   | Velocidad registrada por el radar  |

| radar_id              | VARCHAR | FK → Radar(radar_id) (opcional) |

**Evidencia**

| Atributo       | Tipo    | Descripción                     |
|----------------|---------|---------------------------------|
| imagen         | VARCHAR | Identificador único (PK)        |
| patente_imagen | VARCHAR | Patente detectada por OCR       |
| ratio          | FLOAT   | Similitud entre patentes (0-1)  |
| multa_id       | VARCHAR | FK → Multa(multa_id)            |

### Relaciones

| Relación                        | Cardinalidad |
|---------------------------------|--------------|
| Vehiculo → Multa                | 1 a N        |
| Radar → Multa                   | 1 a N        |
| Multa → Evidencia               | 1 a 0..1     |

````python
#@title Ejercicio 03: clases del modelo lógico


class Vehiculo:
  """Representa un vehículo identificado por su patente."""

  def __init__(self, patente: str) -> None:
    """Inicializa un vehículo con su patente.

    Args:
      patente (str): Patente del vehículo.
    """
    self.patente: str = patente

  def __repr__(self) -> str:
    return f"Vehiculo(patente={self.patente})"


class Radar:
  """Representa un radar ubicado en una posición física."""

  def __init__(self, radar_id: str, ubicacion: str) -> None:
    """Inicializa un radar con su id y ubicación.

    Args:
      radar_id (str): Identificador del radar.
      ubicacion (str): Ubicación física del radar.
    """
    self.radar_id: str = radar_id
    self.ubicacion: str = ubicacion

  def __repr__(self) -> str:
    return (
      f"Radar(radar_id={self.radar_id}, "
      f"ubicacion={self.ubicacion})"
    )


class Evidencia:
  """Representa la evidencia visual asociada a una multa."""

  def __init__(
    self,
    archivo_imagen: str,
    patente_imagen: str,
    ratio: float,
  ) -> None:
    """Inicializa una evidencia con sus datos visuales.

    Args:
      archivo_imagen (str): Nombre del archivo de imagen.
      patente_imagen (str): Patente detectada en la imagen.
      ratio (float): Similitud entre patentes (0 a 1).
    """
    self.archivo_imagen: str = archivo_imagen
    self.patente_imagen: str = patente_imagen
    self.ratio: float = ratio

  def __repr__(self) -> str:
    return (
      f"Evidencia(archivo_imagen={self.archivo_imagen}, "
      f"patente_imagen={self.patente_imagen}, "
      f"ratio={self.ratio})"
    )


class Multa:
  """Representa una multa por exceso de velocidad."""

  def __init__(
    self,
    multa_id: str,
    fecha: str,
    hora: str,
    velocidad_registrada: float,
    velocidad_maxima: int,
    estado_multa: str,
    vehiculo: Vehiculo,
    radar: Radar,
    evidencia: Evidencia | None = None,
  ) -> None:
    """Inicializa una multa con todos sus datos.

    Args:
      multa_id (str): Identificador de la multa.
      fecha (str): Fecha de la infracción.
      hora (str): Hora de la infracción.
      velocidad_registrada (float): Velocidad registrada.
      velocidad_maxima (int): Velocidad máxima permitida.
      estado_multa (str): Estado de la multa.
      vehiculo (Vehiculo): Vehículo infractor.
      radar (Radar): Radar que registró la infracción.
      evidencia (Evidencia | None): Evidencia visual opcional.
    """
    self.multa_id: str = multa_id
    self.fecha: str = fecha
    self.hora: str = hora
    self.velocidad_registrada: float = velocidad_registrada
    self.velocidad_maxima: int = velocidad_maxima
    self.estado_multa: str = estado_multa
    self.vehiculo: Vehiculo = vehiculo
    self.radar: Radar = radar
    self.evidencia: Evidencia | None = evidencia

  def __repr__(self) -> str:
    return (
      f"Multa(multa_id={self.multa_id}, "
      f"patente={self.vehiculo.patente}, "
      f"fecha={self.fecha})"
    )
````

````python
#@title Ejercicio 04: función procesar_fila_csv


def procesar_fila_csv(row: dict) -> Multa:
  """Recibe un diccionario y lo mapea a una instancia Multa.

  Args:
    row (dict): Fila del archivo
      `data/processed/speeding_fines_image.csv`

  Returns:
    Multa: Instancia de la clase Multa con sus relaciones.
  """
  vehiculo: Vehiculo = Vehiculo(
    patente=str(row['patente']),
  )

  radar: Radar = Radar(
    radar_id=str(row['radar_id']),
    ubicacion=str(row['ubicacion']),
  )

  evidencia: Evidencia | None = None
  if row.get('imagen') and str(row['imagen']).strip():
    evidencia = Evidencia(
      archivo_imagen=str(row['imagen']),
      patente_imagen=str(row.get('patente_imagen', '')),
      ratio=float(row.get('ratio', 0.0)),
    )

  return Multa(
    multa_id=str(row['multa_id']),
    fecha=str(row['fecha']),
    hora=str(row['hora']),
    velocidad_registrada=float(row['velocidad_registrada']),
    velocidad_maxima=int(row['velocidad_maxima']),
    estado_multa=str(row['estado_multa']),
    vehiculo=vehiculo,
    radar=radar,
    evidencia=evidencia,
  )
````

# Ejercicio 05

> Puntos: 1


Diseñar el modelo relacional (según el modelo de datos creado en el punto 3) que usaremos con el ORM de SQLAlchemy.

Cada modelo debe incluir:

- Clave primaria

- Relaciones entre tablas

- Sobrescribir el método `__repr__` para un mejor legibilidad del código.





````python
#@title Ejercicio 05: modelos relacionales SQLAlchemy

Base = declarative_base()


class VehiculoORM(Base):
  """Modelo ORM que representa la tabla vehiculos."""

  __tablename__ = 'vehiculos'

  patente = Column(String, primary_key=True)
  multas = relationship(
    'MultaORM',
    back_populates='vehiculo',
  )

  def __repr__(self) -> str:
    return f"VehiculoORM(patente={self.patente})"


class RadarORM(Base):
  """Modelo ORM que representa la tabla radares."""

  __tablename__ = 'radares'

  radar_id = Column(String, primary_key=True)
  ubicacion = Column(String)
  multas = relationship(
    'MultaORM',
    back_populates='radar',
  )

  def __repr__(self) -> str:
    return (
      f"RadarORM(radar_id={self.radar_id}, "
      f"ubicacion={self.ubicacion})"
    )


class EvidenciaORM(Base):
  """Modelo ORM que representa la tabla evidencias."""

  __tablename__ = 'evidencias'

  archivo_imagen = Column(String, primary_key=True)
  patente_imagen = Column(String)
  ratio = Column(Float)
  multa_id = Column(
    Integer,
    ForeignKey('multas.multa_id'),
  )
  multa = relationship(
    'MultaORM',
    back_populates='evidencia',
  )

  def __repr__(self) -> str:
    return (
      f"EvidenciaORM(archivo_imagen={self.archivo_imagen}, "
      f"patente_imagen={self.patente_imagen}, "
      f"ratio={self.ratio})"
    )


class MultaORM(Base):
  """Modelo ORM que representa la tabla multas."""

  __tablename__ = 'multas'

  multa_id = Column(Integer, primary_key=True)
  fecha = Column(Date)
  hora = Column(Time)
  velocidad_registrada = Column(Float)
  velocidad_maxima = Column(Integer)
  estado_multa = Column(String)
  patente = Column(
    String,
    ForeignKey('vehiculos.patente'),
  )
  radar_id = Column(
    String,
    ForeignKey('radares.radar_id'),
  )
  vehiculo = relationship(
    'VehiculoORM',
    back_populates='multas',
  )
  radar = relationship(
    'RadarORM',
    back_populates='multas',
  )
  evidencia = relationship(
    'EvidenciaORM',
    back_populates='multa',
    uselist=False,
  )

  def __repr__(self) -> str:
    return (
      f"MultaORM(multa_id={self.multa_id}, "
      f"patente={self.patente}, "
      f"fecha={self.fecha})"
    )
````

````python
#@title Ejercicio 06: crear engine y tablas

DB_DIR: str = "urban_flow/data/processed"
DB_PATH: str = f"{DB_DIR}/transito.db"

os.makedirs(DB_DIR, exist_ok=True)

engine = create_engine(f"sqlite:///{DB_PATH}", echo=False)
Base.metadata.create_all(engine, checkfirst=True)

print("Base de datos lista.")
print(f"Ubicación: {DB_PATH}")
````

````python
#@title Ejercicio 06: migrar datos desde el CSV

from datetime import datetime

with Session(engine) as session:
  ya_poblada: bool = session.query(MultaORM).count() > 0

if ya_poblada:
  print("Base de datos ya poblada, se omite la migración.")
else:
  with Session(engine) as session:
    with session.begin():

      df_csv: pd.DataFrame = pd.read_csv(CSV_PATH)

      vehiculos: dict = {}
      radares: dict = {}

      for _, row in df_csv.iterrows():

        patente: str = str(row['patente'])
        if patente not in vehiculos:
          vehiculos[patente] = VehiculoORM(patente=patente)
          session.add(vehiculos[patente])

        radar_id_raw = row['radar_id']
        if pd.notna(radar_id_raw):
          radar_id: str = str(radar_id_raw)
          if radar_id not in radares:
            radares[radar_id] = RadarORM(
              radar_id=radar_id,
              ubicacion=str(row['ubicacion']),
            )
            session.add(radares[radar_id])

      session.flush()

      imagenes_vistas: set = set()

      for _, row in df_csv.iterrows():

        imagen: str = str(row.get('imagen', ''))
        if (
          imagen
          and imagen.strip()
          and imagen != 'nan'
          and imagen not in imagenes_vistas
        ):
          imagenes_vistas.add(imagen)
          evidencia: EvidenciaORM = EvidenciaORM(
            archivo_imagen=imagen,
            patente_imagen=str(
              row.get('patente_imagen', '')
            ),
            ratio=float(row.get('ratio', 0.0)),
            multa_id=int(row['multa_id']),
          )
          session.add(evidencia)

        fecha = datetime.strptime(
          str(row['fecha']), '%Y-%m-%d'
        ).date()
        hora_str: str = str(row['hora'])
        fmt: str = (
          '%H:%M:%S' if hora_str.count(':') == 2
          else '%H:%M'
        )
        hora = datetime.strptime(hora_str, fmt).time()

        radar_id_multa = row['radar_id']
        radar_id_valor: str | None = (
          str(radar_id_multa)
          if pd.notna(radar_id_multa)
          else None
        )

        multa: MultaORM = MultaORM(
          multa_id=int(row['multa_id']),
          fecha=fecha,
          hora=hora,
          velocidad_registrada=float(
            row['velocidad_registrada']
          ),
          velocidad_maxima=int(row['velocidad_maxima']),
          estado_multa=str(row['estado_multa']),
          patente=str(row['patente']),
          radar_id=radar_id_valor,
        )
        session.add(multa)

  print("Datos migrados correctamente.")
````

Crear la Base De Datos Vectorial `patente_vectorial` para almacenar los datos utilizando el modelo base OpenClip. Ambas instalaciones se realizaron al inicio del notebook.

Para realizar el almacenamiento vamos a tomar el id del Vehículo y el vector de la imagen.

La idea es que al obtener una imagen podamos obtener por aproximación el id del vehiculo.

Poblar la Base De Datos.



````python
#@title Ejercicio 08: base vectorial con ChromaDB y OpenCLIP
device: str = "cuda" if torch.cuda.is_available() else "cpu"

model, _, preprocess = open_clip.create_model_and_transforms(
  "ViT-B-32", pretrained="openai"
)
model = model.to(device)
model.eval()

chroma_client = chromadb.Client()
coleccion = chroma_client.get_or_create_collection(
  name="patente_vectorial"
)

with Session(engine) as session:
  evidencias = session.query(EvidenciaORM).all()

  for ev in evidencias:
    ruta_imagen: str = os.path.join(IMGS_DIR, ev.archivo_imagen)

    if not os.path.exists(ruta_imagen):
      print(f"No encontrada: {ruta_imagen}")
      continue

    existentes = coleccion.get(ids=[ev.archivo_imagen])
    if existentes["ids"]:
      continue

    multa: MultaORM | None = (
      session.query(MultaORM)
      .filter_by(multa_id=ev.multa_id)
      .first()
    )
    if multa is None:
      continue

    imagen = (
      preprocess(Image.open(ruta_imagen).convert("RGB"))
      .unsqueeze(0)
      .to(device)
    )
    with torch.no_grad():
      vector = model.encode_image(imagen)
      vector = vector / vector.norm(dim=-1, keepdim=True)

    coleccion.add(
      ids=[ev.archivo_imagen],
      embeddings=[vector.squeeze().cpu().tolist()],
      metadatas=[{"patente_vehiculo": multa.patente}],
    )

print(
  f"Base vectorial 'patente_vectorial' poblada "
  f"con {coleccion.count()} elementos."
)
````

# Ejercicio 09

> Puntos: 1

Crear la función `buscar_patente_imagen` que recibe como parámetro una imagen y retorna todos los datos del vehículo.

````python
#@title Ejercicio 09: función buscar_patente_imagen
from sqlalchemy.orm import joinedload

def buscar_patente_imagen(
  ruta_imagen: str,
) -> VehiculoORM | None:
  """Busca el vehículo más similar a una imagen dada.

  Args:
    ruta_imagen: Ruta al archivo de imagen de patente.

  Returns:
    VehiculoORM con sus multas o None si no hay resultados.
  """
  imagen: Image.Image = (
    preprocess(Image.open(ruta_imagen).convert("RGB"))
    .unsqueeze(0)
    .to(device)
  )

  with torch.no_grad():
    vector = model.encode_image(imagen)
    vector = vector / vector.norm(dim=-1, keepdim=True)

  resultado = coleccion.query(
    query_embeddings=[vector.squeeze().cpu().tolist()],
    n_results=1,
  )

  if not resultado["ids"] or not resultado["ids"][0]:
    return None

  patente: str = resultado["metadatas"][0][0]["patente_vehiculo"]

  with Session(engine) as sesion:
    vehiculo: VehiculoORM | None = (
      sesion.query(VehiculoORM)
      .options(joinedload(VehiculoORM.multas))
      .filter(VehiculoORM.patente == patente)
      .first()
    )
  return vehiculo
````
